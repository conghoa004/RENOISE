# Speech & Vocal Noise Reduction with MossFormer2 (ClearVoice)

This notebook provides a modular pipeline for speech enhancement and vocal background noise reduction using **ClearVoice** with the **MossFormer2_SE_48K** model (operating at high-fidelity 48 kHz).

---

### Pipeline Stages
1. **Install Dependencies**: Set up `clearvoice` and audio processing libraries.
2. **Configuration**: Set audio file paths and target model.
3. **Input Validation**: Confirm input file existence and accessibility.
4. **Load Model**: Initialize ClearVoice and load MossFormer2 weights.
5. **Speech Enhancement**: Perform inference and remove background noise.
6. **Export Audio**: Write the enhanced audio to disk.
7. **Audio Preview**: Listen to before/after audio directly in the notebook.

## 2. Configuration & Parameter Setup

Define the input audio file, output destination, and the speech enhancement model to use.

```python
import os
import sys
import time

# ============================================================
# CONFIGURATION
# ============================================================

# Path to the input noisy audio file (e.g., .wav format)
INPUT_FILE = "/content/ghiam.wav"

# Path where the enhanced/cleaned audio will be saved
OUTPUT_FILE = "vocal_clean.wav"

# Pre-trained speech enhancement model (48 kHz monophonic SE)
MODEL = "MossFormer2_SE_48K"

print("Configuration set:")
print(f" - Model      : {MODEL}")
print(f" - Input file : {INPUT_FILE}")
print(f" - Output file: {OUTPUT_FILE}")
```

## 3. Input Validation

Verify that the input audio file exists before loading the model and running inference.

```python
# Verify file existence
if not os.path.isfile(INPUT_FILE):
    raise FileNotFoundError(
        f"[ERROR] Input audio file not found: '{INPUT_FILE}'. "
        "Please verify the path and ensure the file is uploaded."
    )

print(f"[OK] Input file found: {INPUT_FILE}")
```

## 4. Load Pre-trained Model

Initialize the ClearVoice engine and load the `MossFormer2_SE_48K` model weights.
*Note: Pre-trained weights will be downloaded automatically on the first execution.*

```python
from clearvoice import ClearVoice

print("=" * 60)
print(" MossFormer2 48K - Speech Noise Reduction Pipeline")
print("=" * 60)

print("\n[1/3] Loading ClearVoice model...")
print("Note: Pre-trained weights will be downloaded automatically on first run.\n")

start_time = time.time()

enhancer = ClearVoice(
    task="speech_enhancement",
    model_names=[MODEL]
)

print(f"Model successfully loaded in {time.time() - start_time:.2f}s")
```

## 5. Denoise Audio (Inference)

Run the speech enhancement model to suppress background noise from the input audio.

```python
print("[2/3] Processing audio...")
print(f"Input file : {INPUT_FILE}")
print(f"Output file: {OUTPUT_FILE}")
print("\nDenoising in progress, please wait...\n")

start_time = time.time()

output_wav = enhancer(
    input_path=INPUT_FILE,
    online_write=False
)

print(f"Audio processing completed in {time.time() - start_time:.2f}s")
```

## 6. Export Clean Audio

Save the processed waveform to the specified output destination.

```python
print("[3/3] Saving enhanced audio...")

enhancer.write(
    output_wav,
    output_path=OUTPUT_FILE
)

print("\n" + "=" * 60)
print(" PROCESSING COMPLETED")
print("=" * 60)
print("\nEnhanced audio file saved to:")
print(os.path.abspath(OUTPUT_FILE))
print()
```

## 7. Audio Preview (Optional)

Compare the original noisy audio with the enhanced output directly in the notebook.

```python
# Play and compare original vs. enhanced audio in Jupyter / Colab
from IPython.display import Audio, display

if os.path.isfile(INPUT_FILE):
    print("Original (Noisy) Audio:")
    display(Audio(INPUT_FILE))

if os.path.isfile(OUTPUT_FILE):
    print("Enhanced (Clean) Audio:")
    display(Audio(OUTPUT_FILE))
```
